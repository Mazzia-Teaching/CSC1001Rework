### <span style="color:green">Note: Defining functions</span>
- To define functions, we use
        def [function name] (parameters):
            #function operations goes here
        

- To end a function and give back a value, we use 'return'
- Functions do not have to return anything - This is known as a 'void' return type
- Note the lowercase names with \_ between each word
- See simple examples below

```python
#Defining function with parameters
def add(num1, num2):
    return num1 + num2
```

```python
#calling the function, giving it arguments of 5 & 5 to fill the parameter requirements
add(5,5)
```

```python
#What happens if you leave off paremters
add(5)
```

### <span style="color:green">Note: Returning a boolean</span>
- When we return a boolean, it is best to return a boolean expression if we can
- It is redundant to say 'if condition is true, then return true'
- See example below for an example

```python
#Correct way
def is_even_(num):
    return num % 2 == 0 #This will evaluate to true or false, so simply return the boolean expression!
```

```python
#Results are the same, the second method is cleaner
print("5 is even: ", is_even(5))
print("5 is even_:", is_even_(5))

print("6 is even: ", is_even(6))
print("6 is even_: ", is_even_(6))
```

```python
def create_digit_buttons(minValue, maxValue):
    digits = [x for x in range(minValue,maxValue)]
    buttons = [Button(description = str(digit)) for digit in digits]
    return buttons
```

```python
def create_operator_buttons(operators):
    buttons = [Button(description = operator) for operator in operators]
    return buttons
```

```python
def make_flat_list(l): #makes lists of lists one list
    flat_list = [item for sublist in l for item in sublist]
    return flat_list
```

```python
def make_lines(digitButtons, operatorButtons):
    line1 = make_flat_list([digitButtons[7:10],operatorButtons[0:1]])
    line2 = make_flat_list([digitButtons[4:7],operatorButtons[1:2]])
    line3 = make_flat_list([digitButtons[1:4],operatorButtons[2:3]])
    line4 = make_flat_list([digitButtons[0:1],operatorButtons[3:6]])

    return [line1,line2,line3,line4]
```
