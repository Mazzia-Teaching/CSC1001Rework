### <span style="color:green">Note: Statistical Simulation </span>
- Below is an example of using a loop to run a statistical simulation
- The code is examining the difference between rolling 1 twelve-sided die vs rolling 2 six-sided die
- It utilizes the random class to generate random numbers within a particular range
- It loops 'tries' number of times, adding up all the randomly generated numebrs for each die
- Then calculates the average of each by dividing by tries
- The use of the 'time' class is to demonstrate how quickly python can operate
- It grabs the time at the start and end, and calculates how long the function took to operate by getting the difference between the two times

```python
import time
import random
```

```python
def diceTest(tries):
    #initialize all the vars
    counter = 0

    twelveDieCount = 0
    twoSixDieCount = 0

    start = time.time() #Grab time at the beginning

    while counter < tries:
        dice1 = random.randint(1,12)
        dice2 = random.randint(1,6)
        dice3 = random.randint(1,6)
        diceTotal = dice2 + dice3

        twelveDieCount += dice1
        twoSixDieCount += diceTotal

        #print("One \'12 sided die: \'", dice1)
        #print("Two \'6 sided die\'", diceTotal,"\n")
        counter += 1

    print("Average for One  \'12 sided die: \'", twelveDieCount/(tries))
    print("Average for Two \'6 sided die\'", twoSixDieCount/tries)

    end = time.time() #Grab time at the end
    print("As a computer, this took me %.2f seconds to do that %.0f times, let's see YOU try and do that..." % ((end - start),  tries))
```

```python
diceTest(10000000) #This is a billion tries
```

### <span style="color:green">Note: Range function</span>
- The range function has 3 parameters
- range(start,end,step)
- Common in programming, the start parameter is inclusive and the end parameter is exclusive
- Pay attention to each cell, as they demonstrate different variations and default values for the range function

```python
#Here, i is set to the values of the range. i is 0, 1, 2, 3, 4 at each stage of the loop
for i in range(0,5):
    print(i)
```

```python
#There are default values if you leave it blank
for i in range(5): #first value is default of 0
    print(i)
```

```python
#There is a 3rd parameter of 'step' -- How many do you want to go by each time. It defaults to 1
for i in range(0,5,2):
    print(i)
```

```python
#You can also go backwards
for i in range(25,5,-5): #-5 means go back in the sequence by 5
    print(i)
```

## <span style = "color:blue">*Section 2: Traversing strings using for loops*</span>
- In this section, you will learn about using a for loop to traverse a string using the 'in' operator
- This is a very powerful statement that sets the variable you create to each element in the string

```python
myName = "Ted Lasso"
for character in myName: #Using 'in' context of for loop
    print("Character: " + character)
```

### <span style="color:green">Note: Extracting information through a search</span>
- Below, I am using a for loop to grab the first name of an individual
- It adds each character to the new variable firstName
- When it comes across a space, it stops the loops using a break statement

```python
myName = "Ted Lasso"
firstName = ""

for character in myName: #another use of 'in'
    if character == " ": #once we hit the space, stop
        break
    firstName += character

print("firstName: " + firstName)
```

### <span style="color:green">Note: Extracting information through a search</span>
- Below, I am writing a function that takes in a word and a letter and returns how many times that letter appears in the word.
- To do so, I use a for loop to iterate through the string
- Then increment a counter each time the current letter matches the letter we are trying to count
- When the loop is done, function returns the count

```python
def countOccurences(letterToCount, word):
    count = 0
    for letter in word:
        if letter == letterToCount:
            count += 1
    return count
```

```python
countOccurences('a',"Obamna")
```

### <span style="color:green">Note: Using indexes of a string</span>
- Every character in a string is associated with a number. We call this it's 'index'
- **important to note** indexes start at 0. For example the string "Hello":
- H e l l o - the values
- 0 1 2 3 4 - the indexes
- To access a value at it's index, you can use brackets [], see example in Python code below

```python
s = "Hello"
print("s:", s)
print("s[0]:",s[0])
print("s[2:",s[2])
print("s[len(s) - 1]:",s[len(s) - 1]) #last value
print("s[-1]:",s[-1]) #shorthand way of saying last value
print("s[-2]",s[-2]) #second to last value
```

```python
s = "Hello There"
print("s:", s)
print("s[4]:",s[4])
print("s[5]:",s[5]) # Space is a character
print("s[6]:",s[6])
```

```python
#iterating over a string using the indexes:
s = "Yoda"
for i in range(len(s)):
    print(s[i])
```
