```python
def data_sort_1(d):
    return {key : max(value) for key,value in d.items()}

def data_find_1(d):
    return max([val for key,val in d.items()])

def data_sort(d):
    s = data_sort_1(d)
    v = data_find_1(s)
    return [key for key, val in s.items() if val == v]

scores = {'Tony T' : [60,60,60,100],
          'Bobby B' : [59,3,4,5,6],
          'Danny D' :[80,90,67,79],
          'Lonny L' : [64,5,66,100]}

data_sort(scores)
```
