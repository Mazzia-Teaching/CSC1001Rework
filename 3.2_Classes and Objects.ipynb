### <span style="color:green">Note: Defining the class</span>
- Below is a class definition
- The class is called Student
- The self parametes are neccesary throughout a class definition like this

```python
#this is a class
class Student:
    def __init__(self, idNum, firstName, lastName): #how a student is created
        self.idNum = idNum
        self.firstName = firstName
        self.lastName = lastName
        self.grade = 0
        self.onTask = False

    def getFullName(self):
        return self.firstName + " " + self.lastName

    def playMineCraft(self):
        print("Playing minecraft instead of doing work ")
        self.onTask = False

    def doWork(self):
        print("Doing work! My grade has increased by 5")
        self.grade += 5
        self.onTask = True

    def sleep(self):
        print("I am too tired today, ready to sleep")
        self.onTask = False

    def addToGrade(self,amountToAdd):
        self.grade += amountToAdd

    def __str__(self):
        return "\nIdNum: " + str(self.idNum) + "\nFirst Name: " + self.firstName + "\nLast Name: " + self.lastName + "\nGrade: " + str(self.grade) + "\nOn Task: " + str(self.onTask)
```

### <span style="color:green">Note: Calling methods</span>
- Below is an example of calling methods on objects
- Notice how they modify the instance variables of the object created

```python
Ben = Student(123456, "Ben", "SkyWalker")
print("First name:", Ben.firstName)
print("Full name:",Ben.getFullName())
Ben.sleep()
print("On task?", Ben.onTask)
print("Grade", Ben.grade)
Ben.doWork()
print("On task?",Ben.onTask)
print("Grade", Ben.grade)
```

```python
class Character:
    def __init__(self, name, max_health, physical_damage, magic_damage, physical_armour,magic_armour, accuracy):
        self.name = name
        self.maxValues = (max_health, physical_armour, magic_armour)
        self.current_health = max_health
        self.physical_damage = physical_damage
        self.magic_damage = magic_damage
        self.physical_armour = physical_armour
        self.magic_armour = magic_armour
        self.accuracy = accuracy
        
    def take_magic_damage(self, dmgAmount):
        if self.magic_armour > 0: 
            self.magic_armour -= dmgAmount
            if self.magic_armour < 0:
                self.current_health -= abs(self.magic_armour) 
                self.magic_armour = 0 
        else:
            self.current_health -= dmgAmount
        return str(dmgAmount)

    def take_physical_damage(self, dmgAmount):
        if self.physical_armour > 0: 
            self.physical_armour -= dmgAmount
            if self.physical_armour < 0:
                self.current_health -= abs(self.physical_armour)
                self.physical_armour = 0 
        else:
            self.current_health -= dmgAmount
        return str(dmgAmount)

    def take_damage(self,dmgAmount, dmgType):
        if dmgType == "physical":
            return self.name + " Took " + self.take_physical_damage(dmgAmount) + " " + dmgType + " damage"
        elif dmgType == "magic":
            return self.name + " Took " + self.take_magic_damage(dmgAmount) + " " + dmgType + " damage"
        else:
            return "Incorrect dmg type:" + dmgType

    def attack(self, characterToAttack, attackType):
        dmgAmount = 0
        if attackType == "physical":
            dmgAmount = self.physical_damage
        elif attackType == "magic":
            dmgAmount = self.magic_damage
        return characterToAttack.take_damage(dmgAmount,attackType)

    def physical_attack(self,characterToAttack):
        return self.attack(characterToAttack, "physical")

    def magic_attack(self,characterToAttack):
        return self.attack(characterToAttack, "magic")

    def __str__(self):
        return f"{self.name.upper()}\nHP: {self.current_health}\nPhys Arm: {self.physical_armour} | Mag Arm: {self.magic_armour}\nPhys Dmg: {self.physical_damage} | Mag Dmg: {self.magic_damage}"
```
