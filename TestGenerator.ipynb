```python
def every_other_letter(s):
    every_other = ""
    for i in range(0, len(s), 2):
        every_other += s[i]
    return every_other
    
def sub_string(s, terminator):
    if terminator.lower() not in s.lower():
        return "Terminating Character not Found"
    sub_str = ""
    for c in s:
        sub_str += c
        if c.lower() == terminator.lower():
            break
    return sub_str
    
#determines if y is vowel closer look
def examine_y_vowel(s, vowel_count, vowels, y_loc):
    return y_loc == 0 or 'y' == s[-1] or (s[y_loc - 1] not in vowels and s[y_loc + 1] not in vowels)

#determines if y is vowel, looks first if vowel_count == 0
def is_y_vowel(s, vowel_count, vowels, y_loc):
    return True if vowel_count == 0 else examine_y_vowel(s, vowel_count, vowels, y_loc)

def count_basic_vowels(s, vowels):
    vowel_count = 0
    for c in s:
        if c in vowels:
            vowel_count += 1
    return vowel_count

def count_y_vowels(s, vowel_count, vowels):
     for index,letter in enumerate(s):
        if letter == 'y':
            if(is_y_vowel(s, vowel_count, vowels, index)):
                vowel_count += 1
     return vowel_count
def count_vowels(s):
    vowels = "aeiou"
    vowel_count = count_basic_vowels(s,vowels)
    return vowel_count if 'y' not in s else count_y_vowels(s,vowel_count,vowels)
```

```python
#This testing framework was built by Google Gemini and Edited by Mr. Mazzia. Saved me a lot of tedius work! A great use of AI
#Run this to test the full coverage of your code!

import unittest

class TestStringFunctions(unittest.TestCase):
    
    # --- Tests for every_other_letter(s) ---
    def test_every_other_letter_empty(self):
        self.assertEqual(every_other_letter(""), "")
        
        
    def test_every_other_letter_single_char(self):
        self.assertEqual(every_other_letter("A"), "A")
        
    
    def test_every_other_letter_even_length(self):
        self.assertEqual(every_other_letter("abcd"), "ac")
        
        
    def test_every_other_letter_odd_length(self):
        self.assertEqual(every_other_letter("abcde"), "ace")
        
        
    # --- Tests for sub_string(s, terminator) ---
    def test_sub_string_terminator_not_found(self):
        # z not in
        self.assertEqual(sub_string("hello", "z"), "Terminating Character not Found")
        
    def test_sub_string_terminator_case_mismatch(self):
        # 'L' should get sent to lower case and pass
        self.assertEqual(sub_string("hello", "L"), "hel")

    def test_sub_string_terminator_at_start(self):
        #first string sub string
        self.assertEqual(sub_string("hello", "h"), "h")
        
    def test_sub_string_terminator_in_middle(self):
        self.assertEqual(sub_string("hello", "l"), "hel")
        
    def test_sub_string_terminator_at_end(self):
        self.assertEqual(sub_string("hello", "o"), "hello")
        
    def test_sub_string_case_insensitive_match(self):
        # 'L' is in the string at the end. The loop match is case-insensitive, 
        # so it will stop at the first 'l'.
        self.assertEqual(sub_string("helLoL", "L"), "hel")
        
    def test_sub_nothing(self):
        #test with nothing.
        self.assertEqual(sub_string("", "L"), "Terminating Character not Found")    
        
    # --- Tests for count_vowels(s) ---
    def test_count_vowels_empty(self):
        self.assertEqual(count_vowels(""), 0)
        
    def test_count_vowels_basic_only(self):
        self.assertEqual(count_vowels("audio"), 4) # a, u, i, o
        
    def test_count_vowels_no_vowels(self):
        self.assertEqual(count_vowels("bcd"), 0)
        
    def test_count_vowels_y_as_only_vowel(self):
        # y is a vowel because basic count is 0
        self.assertEqual(count_vowels("rhythm"), 1)
        
    def test_count_vowels_y_at_start(self):
        # y_loc == 0
        self.assertEqual(count_vowels("yellow"), 3) # y, e, o
        
    def test_count_vowels_y_at_end(self):
        # 'y' == s[-1]
        self.assertEqual(count_vowels("boy"), 2) # o, y
        
    def test_count_vowels_y_surrounded_by_consonants(self):
        # s[y_loc - 1] not in vowels and s[y_loc + 1] not in vowels
        self.assertEqual(count_vowels("sylvan"), 2) # y, a
        
    def test_count_vowels_y_adjacent_to_vowel(self):
        # y adjacent to basic vowel should not be counted
        self.assertEqual(count_vowels("saying"), 2) # a, i
        
    def test_count_vowels_y_double(self):
        # y adjacent to basic vowel should not be counted
        self.assertEqual(count_vowels("dryly"), 2) # y, y
        
# Run the tests directly in the Jupyter Notebook
if __name__ == '__main__':
    unittest.main(argv=[''], exit=False)
```
