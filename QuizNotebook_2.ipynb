```python
def mystery_a(num):
    return num % 8 == 0
print(mystery_a(12))
```

```python
def mystery_b(str):
    return "Yes" if "psych" in str else "No"
print(mystery_b("Psychology"))
```

```python
def divisible_by_two(num):
    return "Yes" if num % 2 == 0 else "No"
print(f"4 divisible by 2?: {divisible_by_two(4)}")
print(f"5 divisible by 2?: {divisible_by_two(5)}")
```

```python
def mystery_e(num):
    return num % 3 if (num // 2) < 6 else num % 2
def mystery_d(num):
    return num if num % num == 0 else 1
def mystery_c(str):
    x = len(str)
    return mystery_d(x) if x < 10 else mystery_e(x)
print(mystery_c("Hello There"))
```

```python
def mystery_f(age):
    if age >= 18:
        print("You are legally an Adult")
    else:
        print("Sorry kid")
    if age >= 21:
        print("You can legally drink")
    if age >= 25:
        print("You can rent a car for a normal price")
    if age >= 30:
        print("You are UNC")
    if age >= 45:
        print("Mega UNC")
    else:
        print("OK Boomer")
        
mystery_f(30)
```

```python
def mystery_g(num):
    if num % 2 == 0:
        return "Even"
    elif num % 2 == -1:
        return "Prime?"
    else:
        return "Odd"
```

```python
print(mystery_g(-2))
```

```python
def mystery_h(feet):
    if feet >= "5280":
        return "1 mile"
    elif feet >= "3960":
        return "3/4 mile"
    elif feet >= "2640":
        return "1/2 mile"
    elif feet >= "1220":
        return "1/4 mile"
    else:
        return "No mile"

print(mystery_h(2700))
```
