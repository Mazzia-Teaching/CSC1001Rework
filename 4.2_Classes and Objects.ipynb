## <span style = "color:blue">*Section 1: Defining and creating objects from a class*</span>
- In this section you will see a definition of a class and objects that are created from that class.

### <span style="color:green">Note: Defining the class</span>
- Below is a class definition
- The class is called Student
- The self parametes are neccesary throughout a class definition like this

```python
#this is a class
class Student:
    def __init__(self, idNum, firstName, lastName): #how a student is created
        self.idNum = idNum
        self.firstName = firstName
        self.lastName = lastName
        self.grade = 0
        self.onTask = False

    def getFullName(self):
        return self.firstName + " " + self.lastName

    def playMineCraft(self):
        print("Playing minecraft instead of doing work ")
        self.onTask = False

    def doWork(self):
        print("Doing work! My grade has increased by 5")
        self.grade += 5
        self.onTask = True

    def sleep(self):
        print("I am too tired today, ready to sleep")
        self.onTask = False

    def addToGrade(self,amountToAdd):
        self.grade += amountToAdd

    def __str__(self):
        return "\nIdNum: " + str(self.idNum) + "\nFirst Name: " + self.firstName + "\nLast Name: " + self.lastName + "\nGrade: " + str(self.grade) + "\nOn Task: " + str(self.onTask)
```

### <span style="color:green">Note: Creating Objects</span>

- It uses the Student constructor (in the class it is defined by \__init__
- Notice that there is nothing passed to the 'self' parameter -- This is only something neccesary in class definition

```python
Hingle = Student(213456, "Hingle", "McCringleberry")
```

### <span style="color:green">Note: Calling methods</span>
- Below is an example of calling methods on objects
- Notice how they modify the instance variables of the object created

```python
Ben = Student(123456, "Ben", "SkyWalker")
print("First name:", Ben.firstName)
print("Full name:",Ben.getFullName())
Ben.sleep()
print("On task?", Ben.onTask)
print("Grade", Ben.grade)
Ben.doWork()
print("On task?",Ben.onTask)
print("Grade", Ben.grade)
```

### <span style="color:green">Note: Making decisions on two objects</span>
- This is a more realistic use of objects
- We have two objects that have different values
- We call methods on the objects, then check their properties and make decisions using a conditional block that evaluates their attributes

```python
#Lets use the info to make decisions about who I like
myFavoriteStudent = ""
Tyroil = Student(123456, "Tyroil", "Smoochie-Wallace")
Ibrahim = Student(234567, "Ibrahim", "Moizoos")

Tyroil.playMineCraft()
Ibrahim.doWork()

if Tyroil.onTask:
    myFavoriteStudent = Tyroil
elif Ibrahim.onTask:
    myFavoriteStudent = Ibrahim

myFavoriteStudent.addToGrade(5)
print("The best student is: " + myFavoriteStudent.getFullName())

print("\nFavorite Student: " + str(myFavoriteStudent)) #Notice how casting to a string calls our __str__ method!
```
