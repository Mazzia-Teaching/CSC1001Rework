## <span style="color:Blue">*Conditional Blocks*</span>
- if/else structures - very easy to use!

```python
#this is called the conditional statement.
if x > 5:
    print("x is greater than 5") #go inside here if the above criteria is met

else: #Go here if x does not meet the criteria
    print("x is not greater than 5")
```

### <span style="color:green">Note: Control Flow </span>
- Python uses tabbing to define what is 'inside' the if/else statements
- Everything tabbed 1 time to the right immediatley after the conditional statement is what is executed under that condition
- The code picks up after the conditional block on the outer tabbing space

```python
x = int(input("Please enter a number: ")) #Remember to cast numerical input!
if x > 5:
    print("x is greater than 5") #printed if the above condition is met

print("I like x") #executed immediatley after the condition is evaluated.
```

## <span style="color:Blue">*Section 3: Scope*</span>
- In this section you will see how variables created inside of an if statement only live in the context of that if statement
- This can cause errors in your code if not careful

### <span style="color:green">Note: Scoping issues </span>
- Notice below, how we are intializing the variable num1 inside the if statement.
- It is deleted after the conditional block, thus cannot be accessed when we return to the global scope, generating an error

```python
x = 10
if x > 10:
    num1 = 1 #num1 only lives in the 'Scope' of this if statement
else:
    num2 = 2 #num2 only lives in the 'Scope' of this else statement

#I get an error message if I try to access num1, because it was INITIALIZED (originally created) inside the if statement
print(num1)
print(num2)
```

### <span style="color:green">Note: Proper Scoping </span>
- Notice below, how we are intializing the variable num1 & num2 to 0 so that they can be used in the global scope
- Variables created in a global scope can be accessed on the local scope level, not the other way around

```python
#Initialize all the variables before referncing
x = 10
num1 = 0
num2 = 0

if x > 10:
    num1 = 1
else:
    num2 = 2

#No more error message
print(num1)
print(num2)
```

## <span style="color:Blue">*Section 4: Elif*</span>
- 'elif' means "Else If"
- Sometimes we want more than just a simple if/else combination
- This is most common when we are looking to program specific actions to multiple different options. Like a fast food menu
- 'elif' statements only get execute if the previous condition is false

```python
choice = input("Please enter Y for yes, N for no, or M for maybe: ").upper() #Case-insensitive
if choice == "Y":
    print("Yes")
elif choice == "N": #only reached if previous is false
    print("No")
elif choice == "M": #only reached if previous 2 are false
    print("Maybe")
else:
    print("Neither")
```
