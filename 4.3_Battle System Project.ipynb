```python
#The Character Class
import random

class Character:
    def __init__(self, name, maxHealth, physicalDamage, magicDamage, physicalArmour,magicArmour):
        self.name = name
        self.maxValues = (maxHealth, physicalArmour, magicArmour)
        self.currentHealth = maxHealth
        self.physicalDamage = physicalDamage
        self.magicDamage = magicDamage
        self.physicalArmour = physicalArmour
        self.magicArmour = magicArmour

    def takeMagicDamage(self, dmgAmount):
        if self.magicArmour > 0: #take from magic armour first
            self.magicArmour -= dmgAmount
            #account for magic armour being reduced below 0, take from health
            if self.magicArmour < 0:
                self.currentHealth -= abs(self.magicArmour) #take off the proper amount of dmg
                self.magicArmour = 0 #set to 0
        else:
            self.currentHealth -= dmgAmount

        return str(dmgAmount)

    def takePhysicalDamage(self, dmgAmount):
        if self.physicalArmour > 0: #take from phsyical armour first
            self.physicalArmour -= dmgAmount
            #account for physical armour being reduced below 0, take from health
            if self.physicalArmour < 0:
                self.currentHealth -= abs(self.physicalArmour)
                self.physicalArmour = 0 #set to 0
        else:
            self.currentHealth -= dmgAmount

        return str(dmgAmount)

    def takeDamage(self,dmgAmount, dmgType):
        if dmgType == "physical":
            return self.name + " Took " + self.takePhysicalDamage(dmgAmount) + " " + dmgType + " damage"
        elif dmgType == "magic":
            return self.name + " Took " + self.takeMagicDamage(dmgAmount) + " " + dmgType + " damage"
        else:
            return "Incorrect dmg type:" + dmgType

    def attack(self, characterToAttack, attackType):
        dmgAmount = 0
        if attackType == "physical":
            dmgAmount = self.physicalDamage
        elif attackType == "magic":
            dmgAmount = self.magicDamage
        else:
            print("Incorrect attack type:", attackType)

        return characterToAttack.takeDamage(dmgAmount,attackType)

    def physicalAttack(self,characterToAttack):
        return self.attack(characterToAttack, "physical")

    def magicAttack(self,characterToAttack):
        return self.attack(characterToAttack, "magic")

    def resetAfterBattle(self):
        self.currentHealth = self.maxValues[0]
        self.physicalArmour = self.maxValues[1]
        self.magicArmour = self.maxValues[2]
        print("Reset Completed\n Current Health: " + str(self.currentHealth)
        + "\n Physical Armour: " + str(self.physicalArmour)
        + "\n Magic Armour: " + str(self.magicArmour))

    def __str__(self):
        return "Name: " + self.name + "\n Current HP: " + str(self.currentHealth) + "\n Physical Damage: " + str(self.physicalDamage) + "\n Magic Damage: " + str(self.magicDamage) + "\n Physical Armour: " + str(self.physicalArmour) + "\n Magic Armour: " + str(self.magicArmour)
```
