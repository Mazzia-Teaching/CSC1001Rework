## <span style="color:Blue">*Section 1: Compound Conditionals*</span>
- In this section you will learn how to use a combination of more than one boolean statements, using one or more of the following logical joining operators:
    - and -- Both conditions have to be True (also denoted by &&)
    - or -- At least one of the condtions have to be True (also denoted by ||)
    - not -- invertes the result (also denoted by !)
        

### <span style = "color:green">Note: Example using 'and'</span>
- To be president, you must be at least 35 AND be a natural born US citezen
- Run the cell below to test different values

```python
#Request user info
age = int(input("How old are you "))
origin = input("Where were you born? ")

#Use Compound conditional to make determination
if age >= 35 and origin == "U.S.A":
    print("You can run for president.")
else:
    print("You cannot run for president.")
```

### <span style = "color:green">Note: Example using 'or'</span>
- You must be 18 OR above 60in to ride the ride
- Run the cell below to test different values

```python
#Request user info
age = int(input("How old are you "))
height = int(input("How tall are you(inches)"))

#Evaluate on compound conditional
if age >= 18 or height >= 60:
    print("You can ride the ride")
else:
    print("You cannot ride the ride")
```

### <span style="color:green"> Note: Using 'in' to search for a letter </span>
- Write a program that takes a 'yes' or 'no' answer.
- If the user is entering in the answer, it may not be EXACTLY the same. (YES, yes, y, YAS, etc.)
- We can get around this by doing two things to clean up the input
    - Forcing the answer to lowercase
    - Searching for one Unique letter.

- Run the cell below

```python
answer = input("Would you like to continue? ")
answer = answer.lower() #makes it lowercase, standardized
if 'y' in answer: #searches answer for 'y'
    print("Continue")
elif 'n' in answer:
    print("Goodbye")
else:
    print("Do not compute answer: "  + answer)
```

### <span style="color:green">Note: Number Guessing Game Example</span>
- Create number guessing game
- The computer will come up with a number between 1 & 10
- It will then present a new range based on the answer, and give the user another oppurtunity to guess.
- Example:
    - Random number generated is 6
    - User guesses 4
    - User is now given a range of 5 - 10 to guess the number again
- Run the cell below a few times, see if you can get it right!
- Pay close attention to the way things are tabbed and how I was able to create this nested structure

```python
import random
randNumber = random.randint(0,10) #This is how you can get random numbers, pretty easy!

userInput = int(input("Guess of a number between 0 and 10, you have 2 guesses: "))
if userInput == randNumber:
    print("You guessed it:",randNumber)
else:
    if userInput > randNumber:
        print("My number is less than:", userInput)
        userInput = int(input("Guess of a number between 0 and " + str(userInput - 1) + ": "))
        if userInput == randNumber:
            print("You guessed it:",randNumber)
        else:
            print("WRONG, the number is:", randNumber)
    else:
        print("My number is more than:", userInput)
        userInput = int(input("Guess of a number between " + str(userInput + 1) + " and 10: "))
        if userInput == randNumber:
            print("You guessed it:",randNumber)
        else:
            print("WRONG, the number is:", randNumber)
```

## Section 3: Choose Your Own Adventure Logic

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# This dictionary stores all the scenes in your story. 
# "start" is the default first scene.
story_data = {
    "start": {
        "text": "You walk into school on the first day",
        "choices": {
            "Mog the Freshmen": "mog_freshmen",
            "Be kind to the Freshman": "kind_to_freshmen"
        }
    }
}

print("Story engine initialized! Run the next cells to build and play.")
```

```python
def add_choice(choices, widgets):
    num_choices = len(choices) + 1
    c_text = widgets.Text(description=f"Choice {num_choices} ", placeholder="Text the player sees")
    c_dest = widgets.Text(description=f"Dest {num_choices}", placeholder="Node ID this leads to")
    choices.append((c_text, c_dest))
    
def clear_choices(choices):
    for c_text, c_dest in choices:
        c_text = ""
        c_dest = ""

def draw_choices(widgets, choices):
    for c_text, c_dest in choices:
        widgets.HBox([c_text, c_dest])
        
def save_node(b):
    new_choices = {}
    for c_text, c_dest in choices:
        if c_text.value and c_dest.value:
            new_choices[c_text.value] = c_dest.value
    
    # Save to our global dictionary
    story_data[node_id_input.value] = {
        "text": story_text_input.value,
        "choices": new_choices
    }
    
    with creator_out:
        clear_output()
        print(f"✅ Scene '{node_id_input.value}' saved successfully!")
        print(f"Current scenes in memory: {list(story_data.keys())}")
        
        # Clear inputs for the next scene
        node_id_input.value = ""
        story_text_input.value = ""
        clear_choices(choices)
        
    
# Create UI Widgets for the editor
creator_out = widgets.Output()

html_title = widgets.HTML("<h2>📜 Story Creator</h2><p>Create or overwrite a scene (node). Use the exact <b>Node ID</b> you used as a destination in previous choices.</p>")

node_id_input = widgets.Text(description="Node ID:", placeholder="e.g., clear_path")
story_text_input = widgets.Textarea(description="Story Text:", layout=widgets.Layout(width='80%', height='100px'))

# Add Choices
choices = []
add_choice(choices, widgets)

save_btn = widgets.Button(description="Save Scene", button_style='success', icon='save')
add_option = widgets.Button(description="Add Branch", button_style='success', icon='add') 


    
add_option.on_click(add_choice(choices,widgets))
save_btn.on_click(save_node)

# Display the editor interface
rows = [widgets.HBox([c_text, c_dest]) for c_text, c_dest in choices]
editor_ui = widgets.VBox([
    html_title,
    node_id_input,
    story_text_input,
    widgets.HTML("<b>Add Choices (Leave blank for an ending scene):</b>"),
    *rows,
    add_option,
    save_btn,
    creator_out
])

display(editor_ui)
```

```python
player_out = widgets.Output()

def play_scene(node_id):
    with player_out:
        clear_output()
        
        # Check if the player wandered off the map
        if node_id not in story_data:
            display(widgets.HTML(f"<h3 style='color:red;'>Error 404: Scene Not Found</h3>"))
            print(f"The node '{node_id}' hasn't been created yet.")
            print("Go back to the Story Creator to write it!")
            return
            
        node = story_data[node_id]
        
        # Display the story text
        display(widgets.HTML(f"<p style='font-size:16px; line-height:1.5;'>{node['text']}</p>"))
        display(widgets.HTML("<hr>"))
        
        # If there are no choices, it's an ending
        if not node.get("choices"):
            display(widgets.HTML("<b>-- THE END --</b>"))
            restart_btn = widgets.Button(description="Play Again", button_style='info', icon='refresh')
            restart_btn.on_click(lambda b: play_scene("start"))
            display(restart_btn)
            return
            
        # Generate buttons for choices
        buttons = []
        for choice_text, next_node in node["choices"].items():
            btn = widgets.Button(description=choice_text, layout=widgets.Layout(width='auto', min_width='200px'))
            
            # Use a closure to capture the correct next_node for each button
            def make_callback(destination):
                return lambda b: play_scene(destination)
                
            btn.on_click(make_callback(next_node))
            buttons.append(btn)
            
        display(widgets.VBox(buttons))

# Start the game
display(player_out)
play_scene("start")
```
