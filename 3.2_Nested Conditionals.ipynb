### <span style ="color:green">Note: Simple nested conditional example</span>
- This example is not fantastic, as it would be easier ot use an elif statement.
- It is simple so you can learn the basiscs of nesting conditionals before diving into more real world scenarios
- Notice how the second if/else block is indented inside the first else case.
- This is called 'nesting' of a conditional

```python
x = 0
y = 0

if x < y:
    print("x is less than y")
else:
    if x > y:
        print("x is greater than y")
    else:
        print("x and y must be equal")
```

### <span style="color:green">Note: Number Guessing Game Example</span>
- Create number guessing game
- The computer will come up with a number between 1 & 10
- It will then present a new range based on the answer, and give the user another oppurtunity to guess.
- Example:
    - Random number generated is 6
    - User guesses 4
    - User is now given a range of 5 - 10 to guess the number again
- Run the cell below a few times, see if you can get it right!
- Pay close attention to the way things are tabbed and how I was able to create this nested structure

```python
import random
randNumber = random.randint(0,10) #This is how you can get random numbers, pretty easy!

userInput = int(input("Guess of a number between 0 and 10, you have 2 guesses: "))
if userInput == randNumber:
    print("You guessed it:",randNumber)
else:
    if userInput > randNumber:
        print("My number is less than:", userInput)
        userInput = int(input("Guess of a number between 0 and " + str(userInput - 1) + ": "))
        if userInput == randNumber:
            print("You guessed it:",randNumber)
        else:
            print("WRONG, the number is:", randNumber)
    else:
        print("My number is more than:", userInput)
        userInput = int(input("Guess of a number between " + str(userInput + 1) + " and 10: "))
        if userInput == randNumber:
            print("You guessed it:",randNumber)
        else:
            print("WRONG, the number is:", randNumber)
```
