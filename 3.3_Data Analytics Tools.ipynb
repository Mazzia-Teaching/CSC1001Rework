### <span style = "color:blue">*Section 1: Using Pandas to Read CSV and Create DataFrame*</span>
- This section will introduce you to the pandas library
- We will be using a simple 2 column dataset found on <a href = "https://www.kaggle.com/datasets/hedizekri/top-charts-artists-country?select=Top_Charts_Artists_Country.csv"> Kaggle </a>
- You can download the data <a href="https://drive.google.com/file/d/1Xxk1Ow9x_uy2c3tQE50vCznLBRi-B_FH/view?usp=sharing"> here </a> if you are unable to use Kaggle
- This data was made by scraping Spotify data on "Best Songs on Spotify for every year (2000-2023). It contains the top artists, and their country of origin

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
DataFrame = pd.read_csv('Top_Charts_Artists_Country.csv')
data = DataFrame.values #Creates a list structure
DataFrame
```

```python
print(type(data))
print(data)
```

```python
artists = DataFrame["Artist"]
countries = DataFrame["Country"]
```

### <span style = "color:blue">*Section 2: Using Dictionaries to Categorize and Count Data*</span>
- Below, we are going to go through the 2 dimensional structure and create a dictionary structure with:
    - Country as the key
    - List of artists as the value

```python
artistsByCountry = {}
for index,country in enumerate(countries):
    if country not in artistsByCountry:
        artistsByCountry[country] = [artists[index]]
    else:
        artistsByCountry[country].append(artists[index])
```

```python
artistsByCountry
```

```python
topArtistsCount = {}
for country,artistList in artistsByCountry.items():
    topArtistsCount[country] =  len(artistList)
topArtistsCount
```

### <span style="color:green">Note: Sorting Dictionary By Value</span>
- Above, you can see the count laid out above. Wouldn't it be nice to have it sorted by the numerical value?

```python
topArtistsCount_sorted = sorted(topArtistsCount.items(), key=lambda x:x[1],reverse=True)
topArtistsCount_sorted
```

### <span style = "color:blue">*Section 3: Plotting Data*</span>
- Next, we will create a plot of the top countries
    - To do so, we will slice the data to grab the stop 10
    - Then use pyplot to create the plot
    - Using the keys as the x values and the values as the y

```python
artist_countries = [country for country, count in topArtistsCount_sorted[0:10]]
artist_count = [count for country, count in topArtistsCount_sorted[0:10]]
fig,ax = plt.subplots()
ax.bar(artist_countries, artist_count)
#ax.legend()
ax.set_title('Where do the top artists come from?')
ax.set_xlabel("Country of Origin")
ax.set_ylabel("# of Top Artists")
plt.show()
```

### <span style="color:green">Note: Clean up the mess</span>
- Clearly, that isn't a useful chart
- There are many ways to address this. You can increase the spacing, shorten the lables, etc.
- We are going to grab the first 2 letters from each country and use those as the x axis
- This is a good place to start as it is very easy to accomplish. Thanks slicing!

```python
artist_countries = [country[0:2] for country, count in topArtistsCount_sorted[0:10]]
artist_count = [count for country, count in topArtistsCount_sorted[0:10]]
fig,ax = plt.subplots()
ax.bar(artist_countries, artist_count)
ax.set_title('Where do the top artists come from?')
ax.set_xlabel("Country of Origin")
ax.set_ylabel("# of Top Artists")
#ax.legend()
plt.show()
```

### <span style="color:green">Note: Be careful</span>
- Notice something odd? The UK disapeared. This is because the plotting system in Python saw 'Un' from both the United States and United Kingdom and only displayed the 'Un' from the U.S.
- Sometimes we have to go into the data directly and not rely just on algorithms to get it right, so let's change the United Kingdom to Uk in the sorted list

```python
artist_countries = [country[0:2] for country, count in topArtistsCount_sorted[0:10]]
artist_count = [count for country, count in topArtistsCount_sorted[0:10]]
fig,ax = plt.subplots()
ax.bar(artist_countries, artist_count)
ax.set_title('Where do the top artists come from?')
ax.set_xlabel("Country of Origin")
ax.set_ylabel("# of Top Artists")
#ax.legend()
plt.show()
```
