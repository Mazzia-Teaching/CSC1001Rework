## <span style = "color:blue">*Section 1: Tuples*</span>
- This section will introduce you to the tuple data structure
- Tuples are an immutable ways of grouping variables
- They work the same way as lists, but they can't be modified after creation
- Because they can't be changed, they are more space effiecient than a list
- They are best used when you are creating a fixed a structure of variables

### <span style="color:green">Note: Tuple Creation And Access</span>
- To create a tuple, you use ( ), with each value being seperated by a comma.
- To access the values directly, you can use the [] with the apropriate index
- Or, you can 'unpack' the values as shown in the 3rd cell below

```python
#Creating a Tuple of floats
coordinate = (0.0,1.0)
```

```python
#Accessing
print("X:", coordinate[0]) #index starts at 0
print("Y:", coordinate[1])
```

```python
#Unpacking values
x,y = coordinate
print("X:",x)
print("Y:",y)
```

```python
#Tuples can contain any number of values, of varying datatypes
coordinate = (0.0,1.0,1.0,2.0)
x,y,z,w = coordinate
print("X:", x, " Y:", y, " Z:", z, " W:",w)
```

### <span style="color:green">Note: Immutable Nature of Tuples</span>
- Above, you are seeing an example of re-assigning the variable to a new tuple.
- Below, you will see what happens if you try to modify a value of a tuple.

```python
coordinate[1] = 5.0 #generates error
```

## <span style = "color:blue">*Section 2: Lists*</span>
- This section will introduce you to the list data structure
- Lists are mutable
- Lists have many methods you can use with them
- They are best used when the data that is being grouped may change over time, or there will be data added over time

### <span style="color:green">Note: List Creation And Access</span>
- To create a tuple, you use [], with each value being seperated by a comma.
- To access the values directly, you can use the [] with the apropriate index
- To access from the back of the list, use a '-' in front of the index inside the []
- To add to the end of list, use append(element)
- To insert at a specific index, use insert(index,element). This shifts the current item at the index and the subsequent items to the right by 1

```python
#Intialize list
heros = ["Steven", "Garnet", "Peridot", "Pearl"]

#Access the values
print(heros[0])
print(heros[1])
print(heros[2])
print(heros[3])
```

```python
#Replacing an element
heros[3] = "Amethyst"
print(heros)
```

```python
#adding an element
heros.append("Pearl")
print(heros)
```

```python
#Inserting an element
heros.insert(1,'Onion')
print(heros)
```

```python
#Accessing from the back of list
print(heros[-1])
print(heros[-2])
```

### <span style="color:green">Note: Accessing an invalid index</span>
- Below demonstrates what happens if you access an index outside the bounds of the list

```python
heros[6] #generates an error
```

## <span style = "color:blue">*Section 3: List Slicing*</span>
- In this section, you will learn how to use slicing to extract values from a list
- List slicing is an easy and efficient tool to extract values from a list.
- Slicing uses [start:end:step] format, with each having default values. See Lecture 12 slide 7 for more info
- Slicing does not modify the list, it returns a new list containing the sliced values

```python
heros = ['Steven', 'Onion', 'Garnet', 'Peridot', 'Amethyst', 'Pearl']
print(heros)
print("Length of heros: ", len(heros)) #Gets the length of the list
```

```python
print(heros[0:2]) #Slicing elements 0 to 2
```

```python
print(heros[0:1])
print(heros[1:]) #second parameter defaults to end of list if left blank
print(heros[0:len(heros):2]) #Go from 0, to the end, going every other
print(heros[::2]) #First defaults to 0, second to end of list, go by 2 (shorthand of saying the statement directly above)
print(heros[-2:]) #Slice the last 2 values
```

### <span style="color:green">Note: Slicing a string</span>
- Strings work very similar to lists, but they are immutable
- You can slice them to!

```python
name = "Lobos"
print(name[0:3]) #Slicing string
```

## <span style = "color:blue">*Section 5: Traversing Lists*</span>
- In this section, you will learn how to use for loops to traverse a list
- See Lecture 12, slide 17 for detailed information on list traversal

### <span style="color:green">Note: Loop using indexes</span>
- Indexes range from 0 to the length of the list
- Use the index to access the values

```python
for index in range(0,len(heros)):
    print(index,heros[index])
```

### <span style="color:green">Note: Loop using value</span>
- The hero variable is set to each of the values in heros
- The variable name can be anything.

```python
for hero in heros:
    print(hero)
```

### <span style="color:green">Note: Loop using both</span>
- Enumerate generates a tuple pairs of index, value

```python
for index,hero in enumerate(heros):
    print(index,hero)
```

### <span style="color:green">Note: Use loop to extract values meeting a threshold</span>

```python
scores = [95, 90, 85, 48, 90, 40, 46, 57, 98]
passingScores = []
for score in scores:
    if score >= 70:
        passingScores.append(score)
print("PassingScores:", passingScores)
print("Passing Ratio: %.2f" % (len(passingScores)/float(len(scores))))
```

### <span style="color:green">Note: Use List Comprehension to extract values meeting a threshold</span>
- List comprehension condenses the above for loop into 1 statement.
- For more info, See Lecture 12, slide 20.
- The way this reads:
    - I want each score, for each score contained in scores, if that score is above 70.

```python
scores = [95, 90, 85, 48, 90, 40, 46, 57, 98]
passingScores = [score for score in scores if score >= 70]
print("PassingScores:", passingScores)
```
